```python
import pandas as pd
import re

def preprocess_text(text):
    """Convert text to string, remove numbers, URLs, and punctuation."""
    if not isinstance(text, str):  # Ensure it's a string
        text = str(text)

    text = re.sub(r"\d+", "", text)  # Remove numbers
    text = text.lower()  # Convert to lowercase
    text = re.sub(r"http\S+|www\S+|https\S+", "", text, flags=re.MULTILINE)  # Remove URLs
    text = re.sub(r"\s+", " ", text).strip()  # Remove extra spaces
    text = re.sub(r"[^\w\s]", "", text)  # Remove punctuation
    return text

# Ensure the column contains no NaN values before processing
filtered_df["text"] = filtered_df["text"].fillna("")  

# Apply preprocessing
filtered_df["text"] = filtered_df["text"].apply(preprocess_text)

# Remove extra newlines or trailing spaces (if any)
filtered_df["text"] = filtered_df["text"].str.replace(r"\n", " ", regex=True).str.strip()

print(filtered_df["text"].head())
```

```python
import time
import pandas as pd
from deep_translator import GoogleTranslator
from tqdm import tqdm

def split_text(text, max_length=4500):
    """Splits long text into chunks without breaking words."""
    words = text.split()
    chunks = []
    current_chunk = ""

    for word in words:
        if len(current_chunk) + len(word) + 1 <= max_length:
            current_chunk += word + " "
        else:
            chunks.append(current_chunk.strip())
            current_chunk = word + " "
    
    if current_chunk:
        chunks.append(current_chunk.strip())

    return chunks

def translate_with_retries(text, target_lang, retries=3):
    """Translates text with retry logic and error handling."""
    for attempt in range(retries):
        try:
            translated = GoogleTranslator(source="auto", target=target_lang).translate(text)
            time.sleep(1)  # Avoid API rate limits
            return translated
        except Exception as e:
            print(f"Translation error (Attempt {attempt + 1}): {e}")
            time.sleep(2)  # Wait before retrying
    return text

def back_translate(text, lang="fr", max_length=4000):
    """Performs back translation while handling long text and API limits."""
    chunks = split_text(text, max_length)
    translated_chunks = []
    back_translated_chunks = []

    for chunk in tqdm(chunks, desc="Translating Chunks", leave=False):
        translated = translate_with_retries(chunk, lang)
        translated_chunks.append(translated)

        back_translated = translate_with_retries(translated, "en")
        back_translated_chunks.append(back_translated)

    return " ".join(back_translated_chunks)

# Ensure `filtered_df` is defined before this step
if "filtered_df" in globals():
    specific_category = "bdd"  # Change this to the desired category
    category_column = "label"  # Adjust this column name if needed

    # Filter rows based on the specific category
    category_df = filtered_df[filtered_df[category_column] == specific_category].copy()

    # Apply back translation only to the "text" column of the filtered category
    tqdm.pandas(desc="Back Translating Selected Category")
    category_df["text"] = category_df["text"].progress_apply(
        lambda x: back_translate(x) if isinstance(x, str) else x
    )

    # Ensure labels are retained
    if "label" in filtered_df.columns:
        category_df["label"] = filtered_df["label"]  # Retain original labels

    # Append back-translated rows back to the original dataset
    filtered_df = pd.concat([filtered_df, category_df], ignore_index=True)

    # Shuffle dataset to mix original and augmented texts
    filtered_df = filtered_df.sample(frac=1, random_state=42).reset_index(drop=True)
else:
    print("Error: `filtered_df` is not defined. Please ensure it exists before running this code.")
```
